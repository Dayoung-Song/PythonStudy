```python
import pandas as pd
from pandas import DataFrame
from pandas import Series

'''
판다스(pandas)는 파이썬에서 데이터를 쉽게 다룰 수 있게 도와주는 도구이며,
import pandas as pd 코드는 이 도구를 꺼내오는 명령어다.

■ 판다스란?
판다스는 엑셀처럼 표를 다루는 파이썬의 특별한 도구라서, 숫자, 텍스트 등을 표 형태로 정리하고 계산할때 쓴다.

■ 왜 pd라고 줄일까?
pandas 라고 길게 쓰기 귀찮아서 별명을 붙인 것이다. 앞으로도 별명 붙일 때 'as' 사용하자.

■ DataFrame과 Series는 왜 pandas에서 꺼내왔을까?
DataFrame은 행과 열이 있는 2차원 표, Series는 한 줄짜리 표(세로로 길게 나열)하는 pandas의 기능이다.

cf) import pandas as pd 명령어를 사용하면 굳이 DataFrame과 Series를 따로 불러올 필요가 없다.
하지만 별도로 import하지 않으면 pd.Series()처럼 길게 써야 한다.
(별도로 import하면 Series()라고 쓸 수 있음)
'''

pd.DataFrame()  # 엑셀 표 만들기
pd.Series()     # 한 줄 표 만들기  
pd.read_csv()   # 파일 읽기
```
